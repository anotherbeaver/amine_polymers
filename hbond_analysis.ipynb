```python
import MDAnalysis as mda

# Replace 'your_topology.data' and 'trajectory_file.lammpstrj' with your actual file names
u = mda.Universe('data/data.replicated_particles', 'data/dump.equil.lammpstrj', topology_format="DATA", format="LAMMPSDUMP")
```

```python
from MDAnalysis.analysis.hydrogenbonds.hbond_analysis import HydrogenBondAnalysis as HBA
from time import time

start = time()
h = HBA(
    universe=u,
    donors_sel="type 2",
    hydrogens_sel="type 3",
    acceptors_sel="type 4",
    d_a_cutoff=3.5,     # distance cutoff (Angstrom, adjust for LJ units!)
    d_h_a_angle_cutoff=90.0       # angle cutoff (degrees)
)
h.run(backend="multiprocessing", n_workers=4)

end = time()
print(f"HBonds analysis took {end - start:.2f} seconds")
```

```python
import pandas as pd

# lifetimes
# print("Average lifetime (frames):", h.results['life_time'].mean())
# print("All lifetimes:", h.results['life_time'])
results = pd.DataFrame(h.results['hbonds'], columns=['frame', 'donor', 'hydrogen', 'acceptor', 'd-a distance', 'd-h-a angle'])

# save to csv
start = time()

results.to_csv('data/hbonds/hbonds.csv', index=False)
end = time()
print(f"Writing hbonds to file took {end - start:.2f} seconds")
print(results)
```
